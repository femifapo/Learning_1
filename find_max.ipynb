```python
def find_maximum(numbers):
    """
    This function takes a list of numbers and returns the maximum value.
    Arguments:
                numbers (list): A list of numerical values.
    Returns:
                int: The maximum value in the list.
    """
    if not numbers:
        raise ValueError("List of numbers cannot be empty")

    max_value = numbers[0]
    for num in numbers[1:]:
        if num > max_value:
            max_value = num
    return max_value
```

```python
# Example usage
print(find_maximum([3, 7, 2, 9, 5]))  # 9
```
